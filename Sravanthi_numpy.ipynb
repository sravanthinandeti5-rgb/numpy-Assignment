C2. Write a function standardize(x) that takes a 1D NumPy array and returns its z-score standardized version using only NumPy.

Explanation:

Standardization converts the values so that they are on a common scale. NumPy calculates the mean and standard deviation and applies the operation to all elements at once

```python
def standardize(x):
    mean = np.mean(x)
    std = np.std(x)
    return (x - mean) / std
x = np.array([10, 20, 30, 40, 50])
result = standardize(x)
print(result)
```

C4. Given X with shape (100, 5), compute column-wise mean and standard deviation and standardize the matrix in one vectorized expression.

Explanation:
1. axis=0 calculates the mean and standard deviation of each column.
2. X - mean subtracts each column's mean using broadcasting.
3. / std divides each column by its standard deviation.
4. The resulting standardized matrix keeps the same shape (100, 5).

```python
X = np.random.rand(100, 5)
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
X_standardized = (X - mean) / std
print(X_standardized.shape)
```

C5. Implement y_pred = X @ w + b for a feature matrix X, weight vector w and bias b. Then compute MSE using y_true.

Explanation:
X @ w performs matrix multiplication to calculate the weighted values. Then b is added to get the predicted values. MSE calculates the average squared difference between the actual values (y_true) and predicted values (y_pred).

```python
X = np.array([[1, 2],
              [3, 4],
              [5, 6]])
w = np.array([0.5, 1.0])
b = 2
y_true = np.array([4, 7, 10])
y_pred = X @ w + b
mse = np.mean((y_true - y_pred) ** 2)
print("Predicted:", y_pred)
print("MSE:", mse)
```

E1. Per-Column Min–Max Normalization.

Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.

```python
import numpy as np
X = np.array([[70, 800],
              [80, 900],
              [90, 1000]])
min_val = X.min(axis=0)
max_val = X.max(axis=0)
X_normalized = (X - min_val) / (max_val - min_val)
print(X_normalized)
```
